```python
import os

print("=" * 50)
print("DIAGNOSTIC")
print("=" * 50)

print(f"1. Current directory: {os.getcwd()}")

print(f"\n2. Files here:")
for item in os.listdir('.'):
    print(f"   - {item}")

print(f"\n3. Looking for 'src':")
if os.path.exists('src'):
    print("   ✅ src exists in current directory")
    print(f"   Contents: {os.listdir('src')}")
else:
    print("   ❌ src not found here")

print(f"\n4. Looking for 'src' one level up:")
if os.path.exists('../src'):
    print("   ✅ src found one level up")
else:
    print("   ❌ src not found one level up")

print(f"\n5. Looking for 'models' folder:")
if os.path.exists('src/models'):
    print("   ✅ src/models exists")
elif os.path.exists('../src/models'):
    print("   ✅ src/models found one level up")
else:
    print("   ❌ src/models not found")
```
